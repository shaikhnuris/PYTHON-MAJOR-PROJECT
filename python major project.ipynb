```python
# --- IMAGE CRAFT PROJECT (SMART AUTO-EXTENSION CODE) ---
from PIL import Image, ImageOps
import os

def image_craft_app():
    print("=========================================")
    print("         WELCOME TO IMAGE CRAFT          ")
    print("=========================================")
    
    # User se image ka path le rahe hain aur extra quotes ko remove kar rahe hain
    input_path = input("Apni image ka path enter karein (jaise C:/Users/Acer/Pictures/images.jpg): ").strip()
    input_path = input_path.strip('"\'')
    
    # Check kar rahe hain ki file sach mein exist karti hai ya nahi
    if not os.path.exists(input_path):
        print(f"\nError: '{input_path}' naam ki koi file nahi mili!")
        print("Tip: Path mein slashes ko (/) mein badal lein.")
        return

    print("\nKripya apna manchaaha operation chuniye:")
    print("1. Image Resizer and Compressor")
    print("2. Image Format Convertor")
    print("3. Image Effect App (Grayscale / Filter)")
    
    choice = input("Apna option enter karein (1, 2, ya 3): ").strip()
    
    # --- MODULE 1: IMAGE RESIZER AND COMPRESSOR ---
    if choice == '1':
        try:
            width = int(input("Nayi width enter karein (jaise 500): "))
            height = int(input("Nayi height enter karein (jaise 500): "))
            output_path = input("Output image ka naam enter karein (jaise resized): ").strip().strip('"\'')
            
            # Agar user ne extension nahi lagaya, toh automatically .jpg jod denge
            if not output_path.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
                output_path += '.jpg'
            
            with Image.open(input_path) as img:
                img_resized = img.resize((width, height))
                img_resized.save(output_path, quality=75)
                print(f"\nSuccess! Image successfully resize hokar '{output_path}' save ho gayi hai.")
        except Exception as e:
            print(f"Kuch galat ho gaya: {e}")

    # --- MODULE 2: IMAGE FORMAT CONVERTOR ---
    elif choice == '2':
        try:
            target_format = input("Kis format mein convert karna hai? (jaise PNG, JPG, WEBP): ").strip().upper()
            output_path = input("Output file ka naam enter karein (jaise hello.jpg or hello.png): ").strip().strip('"\'')
            
            # Target format ke anusaar automatic extension jod denge (jaise .png ya .jpg)
            ext = f".{target_format.lower()}"
            if not output_path.lower().endswith(ext):
                output_path += ext
            
            with Image.open(input_path) as img:
                # Agar PNG/WEBP se JPEG kar rahe hain toh RGB mode zaroori hai
                if target_format in ["JPG", "JPEG"] and img.mode in ("RGBA", "P"):
                    img = img.convert("RGB")
                    
                img.save(output_path, format=target_format)
                print(f"\nSuccess! Image successfully '{output_path}' mein convert ho gayi hai.")
        except Exception as e:
            print(f"Kuch galat ho gaya: {e}")

    # --- MODULE 3: IMAGE EFFECT APP ---
    elif choice == '3':
        try:
            output_path = input("Effect wali image ka naam enter karein (jaise effect): ").strip().strip('"\'')
            
            # Agar extension nahi lagaya, toh automatically .jpg jod denge
            if not output_path.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
                output_path += '.jpg'
            
            with Image.open(input_path) as img:
                effect_img = ImageOps.grayscale(img)
                effect_img.save(output_path)
                print(f"\nSuccess! Effect successfully lagakar '{output_path}' save kar di gayi hai.")
        except Exception as e:
            print(f"Kuch galat ho gaya: {e}")
            
    else:
        print("Galat option chuna hai! Kripya 1, 2 ya 3 mein se koi ek valid option chunein.")

# Main function ko call kar rahe hain
image_craft_app()
```
